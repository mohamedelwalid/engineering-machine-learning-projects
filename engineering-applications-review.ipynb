#1. Introduction

Modern engineering systems are becoming increasingly complex, and managing them efficiently over time is a challenge. Two areas where this is apparent are energy consumption in buildings and resource usage in manufacturing. They involve dynamic processes with delayed effects. Decisions will directly affect future system states. External uncertainty, such as weather variation, occupancy pattern and supply chain disruptions makes this even harder. Traditional control approaches like rule based systems and PID controllers are built on simplified assumptions that do not hold under real world conditions. This leaves a clear gap for more adaptive methods.

This gap has direct sustainability implications. Buildings alone account for roughly 40% of global energy consumption (IEA, 2023). Inefficient manufacturing contributes to excessive material waste and unnecessary carbon emissions.

Implementation of smarter operational decision making is not a technical improvement but rather a contribution to reducing environmental impact, supporting circular resource use and extending the useful lifetime of products and infrastructure.

Reinforcement learning (RL) offers a principled framework for addressing these problems. RL is modeled as a Markov Decision Process (MDP), which is defined by states, actions, rewards, transition dynamics, and a discount factor γ. RL makes it possible for an agent to learn an optimal policy through direct interaction with an environment, without requiring an explicit system model. RL optimizes for long term cumulative reward and not immediate performance. This aligns naturally with lifecycle thinking. Through reward shaping, it can simultaneously balance competing objectives such as energy efficiency, operational constraints, and user comfort. These properties make RL a compelling and increasingly well studied approach for sustainable optimization across all four phases of the product development lifecycle: design, manufacturing, use, and recycling.

**State space & navigation:**  
The agent navigates a high dimensional discrete state space by iteratively removing elements from the topology. It perfoms an sequential search process through possible design configurations.

**Exploration vs Exploitation:**   
The agent explores different topology configurations during training. The agent will gradually exploit the learned strategies that produce the preferred designs.

**Learning method:**   
The study uses deep reinforcement learning, either DQN or actor-critic based methods.

**Implementation / API:**   

The approach can be implemented using:   
PyTorch (neural network for policy/value function), Custom simulation environment.

Key components:   
Policy network, environment(step, reset).


**Key insight:**   
Reinforcement learning enables efficient exploration of complex design spaces. Through RL, it is possible to achieve either better or comparable performance than traditional optimization methods can.

#Article 2: Operational control
["Wind turbine rotor speed design optimization considering rain erosion based on deep reinforcement learning" (ScienceDirect)](https://www.sciencedirect.com/science/article/pii/S1364032122006724)

**Application context:**
This study applies Deep Reinforcement Learning to optimize wind turbine rotor speed under varying environmental conditions. The goal is minimizing fatigue damage and rain induced erosion while maximizing energy production.

**RL Formulation (MDP):**

Agent: A deep RL policy that determines the optimal rotor speed.

Environment: A physics based simulation of wind turbine dynamics and environmental interactions.

State: A continuous space, including wind speed, rain intensity and possibly the rotor speed or system condition.

Actions: A continuous space, consisting of real time rotor speed adjustments.

Reward: The reward is a hybrid function where fatigue damage gets penalizied while rewarding energy production. This is a trade off between durability and efficiency.

**Initial state:**

The initial state is defined by the environmental conditions at the start of an episode, such as initial wind speed and rain intensity.


**State space & navigation:**
The agent navigates a continuous state space by selecting rotor speeds and observing their impact on fatigue damage and energy output over time.


**Exploration vs Exploitation:**

Exploration is handled through noise added to the policy, allowing the agent to explore different rotor speed strategies while gradually exploiting learned policies

**Learning method:**
The learning method is Deep Deterministic Policy Gradient (DDPG), an actor-critic algorithm where the actor chooses the rotor speed while the critic evaluates.

**Implementation / API:**

The approach can be implemented using:

Pytorch, continuos controll RL framework.

Key components:

Actor network, critic network, replay buffer.

**Key insight:**
This approach enables optimization of long term tradeoffs between structural durability and energy production, which is difficult to achieve using traditional optimization methods.

#3. Most Common RL Applications in the Product Development Lifecycle

Based on the reviewed articles, there seem to be three types of RL applications that keep showing up across the different phases of the product lifecycle.

The most common one is sequential resource allocation and scheduling, especially in manufacturing and recycling. For example, deciding which job to run next, when to do maintenance, or how to plan collection routes. These problems work well with RL because decisions affect future states and the reward is usually something concrete like cost or time.

The second most common one is engineering design optimization, mostly in the design phase. The idea is to use RL to search through large design spaces by trying out different configurations and getting feedback from a simulation. This is useful when the space is complex for traditional optimization methods to handle well.

The third most common one is adaptive personalization, which dominates clearly in the use phase. RL is used here to adjust recommendations or interfaces based on how users actually behave over time instead off just optimizing for immediate clicks or responses.

What these have in common is that they all involve sequential decisions, some degree of uncertainty, and a tradeoff between competing objectives. Which makes sense, since that is basically what RL is designed for.
